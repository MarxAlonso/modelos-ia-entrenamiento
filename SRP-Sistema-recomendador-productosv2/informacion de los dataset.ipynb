```python
import os
import json
import pandas as pd

RUTA_RAW = os.path.join("data", "raw")

def diagnosticar_csv(nombre_archivo, separador=','):
    """Inspecciona estructura, columnas y tipos leyendo solo 500 filas."""
    ruta = os.path.join(RUTA_RAW, nombre_archivo)
    if not os.path.exists(ruta):
        print(f"AVISO: No se encontró {nombre_archivo}")
        return
        
    print("=" * 70)
    print(f"ARCHIVO: {nombre_archivo}")
    print("=" * 70)
    
    # 1. Leemos muestra controlada
    df_muestra = pd.read_csv(ruta, sep=separador, nrows=500)
    
    print("\n--- 1. INFORMACIÓN TÉCNICA (df.info()) ---")
    df_muestra.info()
    
    print("\n--- 2. CONTEO DE VALORES NULOS (df.isnull().sum()) ---")
    print(df_muestra.isnull().sum())
    
    print("\n--- 3. MUESTRA DE FILAS (df.head(2)) ---")
    display(df_muestra.head(2))

def diagnosticar_json(nombre_archivo, max_lineas=2):
    """Inspecciona archivos JSON Lines línea a línea sin saturar RAM."""
    ruta = os.path.join(RUTA_RAW, nombre_archivo)
    if not os.path.exists(ruta):
        print(f"AVISO: No se encontró {nombre_archivo}")
        return

    print("=" * 70)
    print(f"ARCHIVO JSON: {nombre_archivo}")
    print("=" * 70)
    
    muestras = []
    with open(ruta, 'r', encoding='utf-8') as f:
        for i, linea in enumerate(f):
            if i >= max_lineas:
                break
            muestras.append(json.loads(linea))
            
    df_json = pd.DataFrame(muestras)
    print("\n--- INFORMACIÓN TÉCNICA (df.info()) ---")
    df_json.info()
    display(df_json.head(2))
```

```python
# --- Fuentes Crudas (Interacciones y Metadatos) ---
diagnosticar_csv("all_kindle_review .csv")
diagnosticar_csv("ratings_Beauty.csv")
diagnosticar_csv("Amazon_Beauty_Recommendation.csv")
diagnosticar_csv("Books_rating.csv")
diagnosticar_csv("books_data.csv")
diagnosticar_csv("amazon-fashion-800k+-user-reviews-dataset.csv")

diagnosticar_json("Video_Games_5.json")
diagnosticar_json("Electronics_5.json")

# --- Archivos derivados / procesados ---
diagnosticar_csv("preprocessed_kindle_review .csv")
diagnosticar_csv("interacciones.csv")
diagnosticar_csv("productos.csv")
diagnosticar_csv("resenas.csv")
diagnosticar_csv("resenas_traducidas.csv")
```
